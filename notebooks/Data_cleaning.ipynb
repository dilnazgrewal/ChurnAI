```python
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')
df['Total Charges'].dtype
print(df['Total Charges'].isna().sum())
df[df['Total Charges'].isna()][['Tenure Months', 'Monthly Charges', 'Churn Label']]
df.dropna(subset=['Total Charges'], inplace=True)
print(df['Total Charges'].isna().sum())
```

```python
df.groupby('Churn Label')['Churn Reason'].apply(lambda x: x.isna().sum())
```

```python
text_cols = df.select_dtypes(include= str).columns.tolist()
for text in text_cols:
    df[text] = df[text].str.strip().str.title()
    print(f"'{text}' column handled")
```

```python
print(df.loc[df['Churn Value'] > 1].count)
print(df.loc[df['Churn Value'] < 0].count)
```

```python
num_cols = ['Tenure Months', 'Monthly Charges', 'Churn Value', 'Churn Score', 'CLTV']

for col in num_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"\n→ Found {len(outliers)} outliers in {col}")
```
