```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("data_cleaned.xlsx")
```

```python
df.shape
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x='Churn Label',
    hue='Churn Label',
    palette=["#91ACCE", "#B1655C"],
    legend=False
)

plt.title('Distribution of Customer Churn', fontsize=16, fontweight='bold')
plt.xlabel('Churn Status', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

for container in plt.gca().containers:
    plt.bar_label(container)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

*Approximately 73% of customers remained with the company, while 27% churned. Although the majority of customers are retained, the churn rate is significant, highlighting the need for targeted retention strategies to reduce customer loss.*

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x='Contract',
    hue='Churn Label',
    palette=["#91ACCE", "#B1655C"],
    edgecolor='black'
)

plt.title('Customer Churn by Contract Type', fontsize=16, fontweight='bold')
plt.xlabel('Contract Type', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.legend(title='Churn')

for container in plt.gca().containers:
    plt.bar_label(container)

plt.tight_layout()
plt.show()
```

*Customers with month-to-month contracts have the highest churn rate, whereas customers on one-year and two-year contracts are much more likely to stay. Promoting long-term contracts could improve customer retention and reduce churn.*

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='Churn Label',
    y='Tenure Months',
    hue='Churn Label',
    palette=["#91ACCE", "#B1655C"],
    legend=False
)

plt.title('Tenure Months by Customer Churn', fontsize=16, fontweight='bold')
plt.xlabel('Churn Status')
plt.ylabel('Tenure (Months)')

plt.grid(axis='y', alpha = 0.4)
plt.tight_layout()
plt.show()
```

*Customers who churn generally have much shorter tenure than those who remain with the company. This indicates that new customers are at a higher risk of churning, suggesting that improving the onboarding experience and early customer engagement could enhance retention.*

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='Churn Label',
    y='Total Services',
    hue='Churn Label',
    palette=['#4F81BD', '#E74C3C'],
    legend=False
)

plt.title('Total Services by Customer Churn', fontsize=16, fontweight='bold')
plt.xlabel('Churn Status')
plt.ylabel('Number of Services')
plt.grid(axis='y', alpha = 0.4)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))

numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='Blues',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Features', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

*📊 Business Insight*

*The heatmap shows strong positive correlations between Tenure Months and Total Charges (0.83), Monthly Charges and Total Services (0.80), and Total Services and Total Charges (0.80). This indicates that longer-tenured customers tend to accumulate higher total charges, while customers using more services tend to have higher monthly and overall charges.*

*📊 Business Impact*

*The findings highlight the importance of customer retention, as longer customer relationships are associated with greater cumulative customer value. The strong relationship between service adoption and charges also suggests an opportunity to promote relevant additional services, potentially increasing customer value while strengthening customer engagement.*
